### Storm outlines on the reflectivity slider

The map above follows the chosen complex track in longitude and latitude. The figure below draws that same track on the Cartesian radar grid, one volume scan at a time.

Use the slider to change the scan time. Each outline is TITAN's 72-ray polygon: a vertex every 5 degrees, clockwise from north, in Titan X/Y kilometers. The number at the centroid is the simple track. The background is the column-maximum reflectivity.

Set `complexnum` in the next cell to plot a different complex track. `"17"` and `17` select the same track. The plotting code lives in `titan_display.py`, next to this notebook, so this cell only chooses the track and draws it.
